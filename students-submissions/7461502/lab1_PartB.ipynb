### 1.Solve the following two system of linear equations using matrices (find the values of x1 and x2) and 
      determine the number of solutions using the functions. 

```python
import sympy as sp
```

##### 1. Find X1 & X2 Values

```python
# Variables
x1, x2, x3 = sp.symbols('x1 x2 x3')

# Equations
eq1 = sp.Eq(2*x1 + 3*x2 - 4*x3, 6)
eq2 = sp.Eq(x1 - 4*x2, 8)

# Solve
solution1 = sp.solve([eq1, eq2], [x1, x2, x3], dict=True)
solution1
```

##### Find Y1 & Y2

```python
y1, y2, y3 = sp.symbols('y1 y2 y3')

eq3 = sp.Eq(3*y1 - 4*y2 + 5*y3, 10)
eq4 = sp.Eq(-y1 + 2*y2 - 4*y3, 8)

solution2 = sp.solve([eq3, eq4], [y1, y2, y3], dict=True)
solution2
```
